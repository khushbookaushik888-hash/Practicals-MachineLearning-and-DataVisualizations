```python
# Practical 5: Dimensionality Reduction using PCA

# Step 1: Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Step 2: Load dataset
data = load_breast_cancer()

X = data.data
y = data.target

print("Original Dataset Shape:", X.shape)
print("Target Shape:", y.shape)

# Step 3: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Apply PCA
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# Step 5: Display reduced dataset
print("\nReduced Dataset Shape:", X_pca.shape)

print("\nFirst 5 PCA Rows:")
print(X_pca[:5])

# Step 6: Display explained variance
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Explained Variance:",
      sum(pca.explained_variance_ratio_))

# Step 7: Create DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=['Principal Component 1',
             'Principal Component 2']
)

pca_df['Target'] = y

print("\nPCA Dataset:")
print(pca_df.head())

# Step 8: Visualize PCA results
plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=y,
    cmap='viridis',
    alpha=0.7
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA - Dimensionality Reduction")

plt.colorbar(label="Target Class")
plt.show()
```
