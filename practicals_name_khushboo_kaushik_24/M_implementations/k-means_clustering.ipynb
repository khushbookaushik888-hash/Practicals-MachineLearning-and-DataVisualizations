```python

# Step 1: Import libraries

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans


# Step 2: Generate dataset

X, y = make_blobs(
    n_samples=300,
    centers=3,
    n_features=2,
    random_state=42
)

print("Dataset generated successfully")
print("Total records:", X.shape[0])
print("Total features:", X.shape[1])


# Step 3: Visualize original dataset

plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()


# Step 4: Create K-Means model

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)


# Step 5: Train model and assign clusters

y_pred = kmeans.fit_predict(X)


# Step 6: Display cluster labels

print("\nCluster Labels:")
print(y_pred)


# Step 7: Display cluster centers

print("\nCluster Centers:")
print(kmeans.cluster_centers_)


# Step 8: Calculate inertia

print("\nInertia:", kmeans.inertia_)


# Step 9: Visualize clusters

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y_pred,
    cmap='viridis'
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    color='red',
    marker='X',
    s=200,
    label='Centroids'
)

plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
